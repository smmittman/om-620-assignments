## **Question 1 — Required Transformation**

The **fg_mts** subset from Assignment 1 contains finished goods designated as make to stock. Rows labeled **Unknown** in **sku_number** are excluded because their quantities and lead times cannot be assigned to a specific SKU.


```python
known_skus = fg_mts.loc[fg_mts["sku_number"] != "Unknown"]

len(known_skus)
```

This results in 293,462 rows. Which is from 295,343 of fg_mts minus 1,881 with unkown SKUs.


### Summarize order quantity and lead time by SKU

Transactions are grouped by SKU to calculate the minimum, maximum, mean, median, variance, and standard deviation of order quantity. Average lead time is calculated separately for each SKU and is measured in days.


```python
quantity_groups = known_skus["order_quantity"].groupby(known_skus["sku_number"])

lead_time_groups = known_skus["lead_time"].groupby(known_skus["sku_number"])

sku_summary = pd.DataFrame(
    {
        "quantity_min": quantity_groups.min(),
        "quantity_max": quantity_groups.max(),
        "quantity_mean": quantity_groups.mean(),
        "quantity_median": quantity_groups.median(),
        "quantity_variance": quantity_groups.var(),
        "quantity_std": quantity_groups.std(),
        "average_lead_time": lead_time_groups.mean(),
    }
)

sku_summary.head()
```

```python
len(sku_summary)
```

The summary contains one row for each of the 400 identifiable finished-goods SKUs designated as make to stock, matching the count in Assignment 1, Question 5. The quantity statistics summarize individual transactions, and average lead time is measured in days. The 1,881 transactions with unknown SKU numbers were excluded because they could not be reliably assigned to a specific SKU.


## **Question 2 - Safety Stock Calculation**

Safety stock is calculated by multiplying the standard normal z-score by the SKU's order quantity standard deviation and the square root of its average lead time. The norm.ppf() method converts the 75%, 90%, and 95% service levels into standard normal z-scores (SciPy, n.d.). The results are stored as z_75, z_90, and z_95 and their values are verified for each service level. Each z-score is the same for all 400 SKUs at that service level, so it only needs to be calculated once. Using the named variables in the safety stock formulas makes it clear which service level each column uses and avoids repeating the norm.ppf() calls.


```python
z_75 = norm.ppf(0.75)
z_90 = norm.ppf(0.90)
z_95 = norm.ppf(0.95)
```

```python
z_75
```

```python
z_90
```

```python
z_95
```

#### Calculate safety stock columns using the formula $SS = z_{\alpha} \times \sigma_D \times \sqrt{L}$


```python
sku_summary["ss_75"] = (
    z_75 * sku_summary["quantity_std"] * (sku_summary["average_lead_time"] ** 0.5)
)
sku_summary["ss_90"] = (
    z_90 * sku_summary["quantity_std"] * (sku_summary["average_lead_time"] ** 0.5)
)
sku_summary["ss_95"] = (
    z_95 * sku_summary["quantity_std"] * (sku_summary["average_lead_time"] ** 0.5)
)
```

## **Side Quest**


```python
from scipy.stats.mstats import mquantiles


empirical_stock = sku_summary.copy()

for sku in empirical_stock.index:
    quantities = known_skus.loc[known_skus["sku_number"] == sku, "order_quantity"]

    quantiles = mquantiles(quantities, prob=[0.75, 0.90, 0.95])

    average = quantities.mean()
    lead_time = empirical_stock.loc[sku, "average_lead_time"]


    stock_75 = (quantiles[0] - average) * lead_time**0.5
    stock_90 = (quantiles[1] - average) * lead_time**0.5
    stock_95 = (quantiles[2] - average) * lead_time**0.5

    if stock_75 < 0:
        stock_75 = 0

    if stock_90 < 0:
        stock_90 = 0

    if stock_95 < 0:
        stock_95 = 0

    empirical_stock.loc[sku, "empirical_75"] = stock_75
    empirical_stock.loc[sku, "empirical_90"] = stock_90
    empirical_stock.loc[sku, "empirical_95"] = stock_95

display(empirical_stock[["empirical_75", "empirical_90", "empirical_95"]])
```
